```python

import base64
import base64
import os

from langchain.agents import create_agent
from langchain_core.messages import HumanMessage
from langchain_core.tools import tool
from langchain_openai import ChatOpenAI


qwen = ChatOpenAI(
    api_key=os.getenv("DASHSCOPE_API_KEY"),
    base_url="https://dashscope.aliyuncs.com/compatible-mode/v1",
    model="qwen3.6-plus",
    temperature=0.7,
    max_tokens=1000,
    timeout=30,
)

@tool
def get_weather(location: str) -> str:
    """Get the weather in a given location."""
    return f"Current weather in {location} is sunny"


agent = create_agent(
    model=qwen,
    tools=[get_weather],
)

# 阻塞式调用
response = agent.invoke(
    input={"messages": [{"role": "user", "content": "南京今天天气如何"}]},
)
print(response)
# 流式调用
response = agent.stream(
    input={"messages": [{"role": "user", "content": "你是谁"}]},
    stream_mode="messages" #流式读的模式
)
for token,metadata in response:
    if token.content: #如果有content
        print(token.content,end="",flush=True)
```

```python
# 我们收到加载环境变量中的base_url和api_key
import osn
from langchain.chat_models import init_chat_model
from pydantic import BaseModel
from langchain_core.messages import HumanMessage
from langchain.agents import create_agent

# 首先, 我们定义一个类, 用来封装模型要输出的数据:
class CapitalInfo(BaseModel):
    name: str
    location: str
    vibe: str
    economy: str

api_key = os.getenv("DASHSCOPE_API_KEY")

model = init_chat_model(
    model="qwen-max",  # 模型名称，这里可以自定义，我们用的是阿里的qwen-max
    model_provider="openai",  # 如果是Langchain不支持的模型，需要指定模型提供商  # 兼容openai，所以这里填openai
    base_url="https://dashscope.aliyuncs.com/compatible-mode/v1",
    api_key=api_key
)

# 我们可以创建智能体时设置结构化输出的格式，LangChain会自动帮我们完成提示词改造和响应结果解析。

agent = create_agent(
    model=model,
    system_prompt="你是一个科幻作家，根据用户的要求创建一个太空之都。",
    response_format=CapitalInfo  # 设置结构化输出的格式
)

response = agent.invoke(
    input={"messages": [HumanMessage(content="月球的首都是什么?")]},
)

# 输出结果
print(response['structured_response'])

```
